# **TASK--> 1**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('11.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

beta = 60
bright = cv2.convertScaleAbs(image, beta=beta)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.axis('off')
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(bright)
plt.axis('off')
plt.title('Brightened Image')

plt.show()
```

```python
alpha=1.5
beta=40

contrast = np.clip(alpha*image + beta, 0,255).astype(np.uint8)

plt.figure(figsize=(10, 5))

plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(image)
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Contrast")
plt.imshow(contrast)
plt.axis('off')
```

```python
gamma = 2.2

gamma = np.power(image/255.0,1/gamma)*255.0
gamma = np.clip(gamma,0,255).astype(np.uint8)

plt.figure(figsize=(10, 5))

plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(image)
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Gamma Correction")
plt.imshow(gamma)
plt.axis('off')
```

```python
negative = 255 - image

plt.figure(figsize=(10, 5))

plt.subplot(1,2,1)
plt.title("Original Image")
plt.imshow(image)
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Negative Image")
plt.imshow(negative)
plt.axis('off')
```

# Addition On Images
## This brightens the resultant image, enhancing areas where both images have color, while potentially saturating bright regions.


```python
img1 = cv2.imread('10.jpg')
img2 = cv2.imread('4.jpg')

img1 = cv2.resize(img1, (img2.shape[1], img2.shape[0]))
add = cv2.add(img1, img2)

cv2.imwrite('add.jpg', add)

plt.figure(figsize=(10, 5))

plt.subplot(1,3,1)
plt.title("Image 1")
plt.imshow(cv2.cvtColor(img1,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,3,2)
plt.title("Image 2")
plt.imshow(cv2.cvtColor(img2,cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1,3,3)
plt.title("Addition")
plt.imshow(add)
plt.axis('off')
```

# Subtraction On Images
## This darkens the resultant image, revealing differences where the first image is lighter than the second.

```python
import cv2
import matplotlib.pyplot as plt

img1 = cv2.imread('10.jpg')
img2 = cv2.imread('4.jpg')

img1 = cv2.resize(img1, (img2.shape[1], img2.shape[0]))
sub = cv2.subtract(img1, img2)

cv2.imwrite('sub.jpg', sub)

plt.figure(figsize=(10, 5))

plt.subplot(1, 3, 1)
plt.title("Image 1")
plt.imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Image 2")
plt.imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Subtraction")
plt.imshow(cv2.cvtColor(sub, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

# Multiplication On Images
## This increases contrast and can create darker images, as it dims colors where either image has low intensity.

```python
import cv2
import matplotlib.pyplot as plt

img1 = cv2.imread('10.jpg')
img2 = cv2.imread('4.jpg')

img1 = cv2.resize(img1, (img2.shape[1], img2.shape[0]))
mul = cv2.multiply(img1, img2)

cv2.imwrite('mul.jpg', mul)

plt.figure(figsize=(10, 5))

plt.subplot(1, 3, 1)
plt.title("Image 1")
plt.imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Image 2")
plt.imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Multiplication")
plt.imshow(cv2.cvtColor(mul, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

# Division On Images
## This can enhance bright areas and create a lighter image but may introduce artifacts if the divisor has low values.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img1 = cv2.imread('10.jpg')
img2 = cv2.imread('4.jpg')

img1 = cv2.resize(img1, (img2.shape[1], img2.shape[0]))
img2 = np.where(img2 == 0, 1, img2)
div = cv2.divide(img1, img2)

cv2.imwrite('div.jpg', div)

plt.figure(figsize=(10, 5))

plt.subplot(1, 3, 1)
plt.title("Image 1")
plt.imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Image 2")
plt.imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Division")
plt.imshow(cv2.cvtColor(div, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

# **TASK--> 3**

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Read the input image and background image
img = cv2.imread("5.png")
img2 = cv2.imread("11.jpg")

# Resize the background image to match the input image size
img2 = cv2.resize(img2, (img.shape[1], img.shape[0]))

# Convert the input image to HSV color space
copy = np.copy(img)
copy = cv2.cvtColor(copy, cv2.COLOR_BGR2HSV)

# Display the HSV image (converted back to RGB for correct color representation)
plt.imshow(cv2.cvtColor(copy, cv2.COLOR_HSV2RGB))
plt.title("HSV Image")
plt.axis('off')
plt.show()
```

```python

# Define the lower and upper bounds for the green color in HSV
lower_green = np.array([35, 100, 100])
upper_green = np.array([85, 255, 255])

# Create a mask for the green area
mask = cv2.inRange(copy, lower_green, upper_green)

# Display the mask
plt.imshow(mask, cmap="gray")
plt.title("Mask")
plt.axis('off')
plt.show()
```

```python

# Invert the mask
mask_invert = cv2.bitwise_not(mask)

# Display the inverted mask
plt.imshow(mask_invert, cmap="gray")
plt.title("Inverted Mask")
plt.axis('off')
plt.show()
```

```python
# Extract the foreground and background using the masks
fg = cv2.bitwise_and(img, img, mask=mask_invert)
bg = cv2.bitwise_and(img2, img2, mask=mask)

plt.imshow(cv2.cvtColor(bg, cv2.COLOR_BGR2RGB))
plt.title("BG Image")
plt.axis('off')
plt.show()
```

```python

# Combine the foreground and background
final = cv2.add(fg, bg)

# Display the final image (converted to RGB for correct color representation)
plt.imshow(cv2.cvtColor(final, cv2.COLOR_BGR2RGB))
plt.title("Final Image")
plt.axis('off')
plt.show()
```

# **TASK--> 4**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

frame1 = cv2.imread('f1.jpg')
frame2 = cv2.imread('f2.jpg')

gray1 = cv2.cvtColor(frame1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(frame2, cv2.COLOR_BGR2GRAY)

gray2 = cv2.resize(gray2, (gray1.shape[1], gray1.shape[0]))

diff = cv2.absdiff(gray1, gray2)

_, thresh = cv2.threshold(diff, 30, 255, cv2.THRESH_BINARY)

kernel = np.ones((5,5), np.uint8)
dilated = cv2.dilate(thresh, kernel, iterations=2)

plt.figure(figsize=(12,7))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(frame1, cv2.COLOR_BGR2RGB))
plt.title('Frame 1')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(frame2, cv2.COLOR_BGR2RGB))
plt.title('Frame 2')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(dilated, cmap='gray')
plt.title('Motion Detected')
plt.axis('off')

plt.show()
```
